```python
from pathlib import Path

import numpy as np
import trimesh


BASE_DIR = Path.cwd()

TOWER_COLLISION_PATH = BASE_DIR / "Tower.STL"
TOWER_VISUAL_PATH = BASE_DIR / "Tower_visual.STL"

FOOT1_PATH = (
    BASE_DIR
    / "candidate_output"
    / "foot1_candidates.npz"
)

FOOT2_PATH = (
    BASE_DIR
    / "candidate_output"
    / "foot2_candidates.npz"
)


def load_stl(path: Path) -> trimesh.Trimesh:
    """
    读取STL。

    某些文件可能被读取为Scene，
    这里统一合并成一个Trimesh。
    """

    geometry = trimesh.load(
        path,
        process=False,
    )

    if isinstance(geometry, trimesh.Scene):
        geometry = geometry.dump(
            concatenate=True
        )

    if not isinstance(geometry, trimesh.Trimesh):
        raise TypeError(
            f"无法识别为三角网格：{path}"
        )

    return geometry


def print_mesh_information(
    name: str,
    mesh: trimesh.Trimesh,
) -> None:
    """打印STL几何信息。"""

    bounds_min = mesh.bounds[0]
    bounds_max = mesh.bounds[1]

    bounding_box_center = (
        bounds_min + bounds_max
    ) / 2.0

    origin_inside_aabb = np.all(
        bounds_min <= 0.0
    ) and np.all(
        0.0 <= bounds_max
    )

    print()
    print("=" * 60)
    print(name)
    print("=" * 60)

    print("顶点数量：", len(mesh.vertices))
    print("三角面数量：", len(mesh.faces))

    print("坐标最小值：", bounds_min)
    print("坐标最大值：", bounds_max)

    print("包围盒尺寸：", mesh.extents)
    print("包围盒中心：", bounding_box_center)
    print("几何质心：", mesh.centroid)

    print(
        "世界原点是否位于包围盒内：",
        origin_inside_aabb,
    )

    print(
        "包围盒中心相对原点偏移：",
        bounding_box_center,
    )


tower_collision_mesh = load_stl(
    TOWER_COLLISION_PATH
)

tower_visual_mesh = load_stl(
    TOWER_VISUAL_PATH
)


print_mesh_information(
    "Tower.STL",
    tower_collision_mesh,
)

print_mesh_information(
    "Tower_visual.STL",
    tower_visual_mesh,
)
```

```python
collision_center = (
    tower_collision_mesh.bounds[0]
    + tower_collision_mesh.bounds[1]
) / 2.0

visual_center = (
    tower_visual_mesh.bounds[0]
    + tower_visual_mesh.bounds[1]
) / 2.0


print()
print("=" * 60)
print("两个STL对比")
print("=" * 60)

print(
    "包围盒中心差：",
    visual_center - collision_center,
)

print(
    "尺寸差：",
    tower_visual_mesh.extents
    - tower_collision_mesh.extents,
)

print(
    "尺寸比例：",
    tower_visual_mesh.extents
    / tower_collision_mesh.extents,
)
```
